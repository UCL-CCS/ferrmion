```python
import pickle
from pathlib import Path
import rustworkx as rx
from ferrmion.optimize.bonsai import bonsai_algorithm
from ferrmion.hamiltonians import molecular_hamiltonian
from ferrmion.utils import fermionic_to_sparse_majorana
from ferrmion.optimize.topphatt import topphatt

def water_integrals():
    folder = Path.cwd().joinpath(Path("../../../python/tests/"))
    with open(folder.joinpath("./data/water_1e.pkl"), 'rb') as file:
        ones = pickle.load(file)

    with open(folder.joinpath("./data/water_2e.pkl"), 'rb') as file:
        twos = pickle.load(file)
    return (ones, twos)
ones, twos = water_integrals()
twos *= 0.5

emerald_edges = [(0, 1),
    (0, 4),
    (3, 2),
    (3, 4),
    (3, 9),
    (5, 1),
    (5, 4),
    (5, 6),
    (5, 11),
    (8, 2),
    (8, 7),
    (8, 9),
    (8, 16),
    (10, 4),
    (10, 9),
    (10, 11),
    (10, 18),
    (12, 11),
    (12, 13),
    (12, 20),
    (15, 7),
    (15, 14),
    (15, 16),
    (15, 23),
    (17, 16),
    (17, 18),
    (17, 25),
    (18, 26),
    (19, 11),
    (19, 18),
    (19, 20),
    (19, 27),
    (21, 13),
    (21, 20),
    (21, 29),
    (22, 14),
    (22, 23),
    (24, 16),
    (24, 23),
    (24, 25),
    (24, 32),
    (25, 26),
    (26, 27),
    (26, 34),
    (28, 20),
    (28, 27),
    (28, 29),
    (28, 36),
    (30, 29),
    (30, 38),
    (31, 23),
    (31, 32),
    (33, 25),
    (33, 32),
    (33, 34),
    (33, 41),
    (34, 42),
    (35, 27),
    (35, 34),
    (35, 36),
    (35, 43),
    (36, 37),
    (37, 38),
    (37, 45),
    (40, 32),
    (40, 39),
    (40, 41),
    (40, 46),
    (41, 42),
    (43, 49),
    (44, 36),
    (44, 43),
    (44, 45),
    (44, 50),
    (47, 41),
    (47, 46),
    (47, 48),
    (47, 51),
    (49, 50),
    (49, 53),
    (52, 48),
    (52, 51),
    (52, 53)
]

#rustworkx graph from edges (the actual input for bonsai)

emerald_graph_rx = rx.PyGraph()
num_qubits = max(max(pair) for pair in emerald_edges) + 1
emerald_graph_rx.add_nodes_from(range(num_qubits))
emerald_graph_rx.add_edges_from_no_data(emerald_edges)

bonsai_emerald_truncated = bonsai_algorithm(
    graph = emerald_graph_rx, 
    homogenous = False, 
    max_nodes = 14
)


bonsai_ham = molecular_hamiltonian(
    encoding = bonsai_emerald_truncated, 
    one_e_coeffs = ones, 
    two_e_coeffs = twos, 
    constant_energy = 0
)

bonsai_topphatt = topphatt(
    majorana_ham = fermionic_to_sparse_majorana(((ones,"+-"), (twos, "++--"))),
    tree = bonsai_emerald_truncated
)
```
