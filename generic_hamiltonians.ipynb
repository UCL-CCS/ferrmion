```rust
use ndarray::{Axis, Zip};
// use ndarray::{azip, concatenate, Axis, Zip};
use itertools::iproduct;
use numpy::ndarray::{ArrayView, s, ArrayView1, ArrayView2, ArrayView4, ArrayD, arr1, arr2};
use numpy::Complex64;
use std::collections::HashMap;
use ndarray::*;
```

```rust
let mut ipowers: Array1<usize> = ndarray::arr1(&[0,1,2,3,4,5,6,7]);
```

```rust
Zip::from(ipowers.slice_mut(s![..;2])).for_each(|i| {*i = (*i + 2) % 4})
```

```rust
let mut ipowers: Array2<usize> = ndarray::arr2(&[[0,1,2,3],[4,5,6,7]]);
println!("{:#}",ipowers);
ipowers.map_inplace(|i| {*i= *i + 2});
println!("{:#}",ipowers);
```

```rust
let mut ipowers: Array2<usize> = ndarray::arr2(&[[0,1,2,3],[4,5,6,7]]);

fn negate_sign(val: &mut usize) {
    *val = (*val +2) % 4
}

println!("{:#}",ipowers);
ipowers.slice_mut(s![..;2,..]).map_inplace(negate_sign);
println!("{:#}",ipowers);
```
