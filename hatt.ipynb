$=\frac{1}{4}(m_{2j}m_{2k} + m_{2j+1}m_{2k+1} + i m_{2j}m_{2k+1} - i m_{2j+1}m_{2k})$

```python
import numpy as np


def fermion_to_majorana(ones, twos):
    n_modes = ones.shape[0]
    majorana_ones = np.zeros((2 * n_modes, 2 * n_modes), dtype=np.complex64)
    majorana_ones.shape
    # first term
    majorana_ones[::2, ::2] = 0.25 * ones
    # second term
    majorana_ones[1::2, 1::2] = -0.25 * ones
    # third tem
    majorana_ones[::2, 1::2] = 0.25j * ones
    # fourth
    majorana_ones[1::2, ::2] = -0.25j * ones

    majorana_twos = np.zeros(
        (2 * n_modes, 2 * n_modes, 2 * n_modes, 2 * n_modes), dtype=np.complex64
    )
    scale_dict = {(0, 0): 1, (1, 1): -1, (0, 1): -1j, (1, 0): -1j}
    for i in range(2):
        for j in range(2):
            for m in range(2):
                for l in range(2):
                    majorana_twos[i::2, j::2, m::2, l::2] = (
                        scale_dict[(i, j)] * np.conj(scale_dict[(m, l)]) * 1 / 16 * twos
                    )
    return majorana_ones, majorana_twos
```
